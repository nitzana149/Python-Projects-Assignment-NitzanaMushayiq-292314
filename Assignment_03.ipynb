```python
n = int(input("Enter a number: "))

total = 0

while n > 0:
    digit = n % 10
    total = total + digit
    n = n // 10

print("Sum of digits =", total)
```

```python
n = int(input("Enter a number: "))

reverse = 0

while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n = n // 10

print("Reversed number =", reverse)
```

```python
total = 0

while True:
    n = int(input("Enter a number (0 to stop): "))

    if n == 0:
        break

    total = total + n

print("Sum of numbers =", total)
```
