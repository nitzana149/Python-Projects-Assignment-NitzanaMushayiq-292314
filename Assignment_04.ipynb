```python
numbers = []

for i in range(10):
    num = int(input("Enter a number: "))
    numbers.append(num)

largest = numbers[0]
smallest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

    if num < smallest:
        smallest = num

print("List =", numbers)
print("Largest element =", largest)
print("Smallest element =", smallest)
```

```python
numbers = [10, 20, 10, 30, 20, 40]

unique = []

for num in numbers:
    if num not in unique:
        unique.append(num)

print("Original list =", numbers)
print("List after removing duplicates =", unique)
```

```python
numbers = [12, 7, 5, 18, 20, 9]

even = []
odd = []

for num in numbers:
    if num % 2 == 0:
        even.append(num)
    else:
        odd.append(num)

print("Original list =", numbers)
print("Even numbers =", even)
print("Odd numbers =", odd)
```
