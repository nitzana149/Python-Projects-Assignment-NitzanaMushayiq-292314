```python
def addition(a, b):
    return a + b

def subtraction(a, b):
    return a - b

def multiplication(a, b):
    return a * b

def division(a, b):
    return a / b

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

print("Addition =", addition(num1, num2))
print("Subtraction =", subtraction(num1, num2))
print("Multiplication =", multiplication(num1, num2))

if num2 != 0:
    print("Division =", division(num1, num2))
else:
    print("Division is not possible")
```

```python
def celsius_to_fahrenheit(c):
    return (c * 9/5) + 32

def fahrenheit_to_celsius(f):
    return (f - 32) * 5/9

def celsius_to_kelvin(c):
    return c + 273.15

c = float(input("Enter temperature in Celsius: "))
f = float(input("Enter temperature in Fahrenheit: "))

print("Celsius to Fahrenheit =", celsius_to_fahrenheit(c))
print("Fahrenheit to Celsius =", fahrenheit_to_celsius(f))
print("Celsius to Kelvin =", celsius_to_kelvin(c))
```

```python
%%writefile module_factorial.py

def factorial(n):
    fact = 1

    for i in range(1, n + 1):
        fact = fact * i

    return fact
```

```python
import module_factorial

num = int(input("Enter a number: "))

result = module_factorial.factorial(num)

print("Factorial =", result)
```

```python
def check_prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

num = int(input("Enter a number: "))

if check_prime(num):
    print("Prime number")
else:
    print("Not a prime number")
```

```python
def string_reverse(text):
    reverse = ""

    for i in range(len(text) - 1, -1, -1):
        reverse = reverse + text[i]

    return reverse

string = input("Enter a string: ")

result = string_reverse(string)

print("Reversed string =", result)
```
