```python
n = int(input("Enter the value of n: "))

total = 0

for i in range(1, n + 1):
    total = total + i ** 2

print("Sum of the series =", total)
```

```python
n = int(input("Enter a number: "))

if n <= 1:
    print("Not a prime number")
else:
    for i in range(2, n):
        if n % i == 0:
            print("Not a prime number")
            break
    else:
        print("Prime number")
```
